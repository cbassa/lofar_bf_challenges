# Find a pulsar in LOFAR dynamic spectra

Pulsar signals stand out in radio astronomy due to their periodicity and frequency dependent dispersion.

In this challenge you are tasked with determining the spin period and dispersion measure (DM) of a mystery pulsar in a 60 second dynamic spectrum taken with the HBA antennas of the LOFAR Superterp stations (observation L2039118). 

Using standard python libraries such as `numpy`, `scipy`, `matplotlib` and `h5py`, you'll learn to read LOFAR data, search for periodic signals using power spectra, and search over DM.

### Hints and tips

* Dispersion of a radio signal with dispersion measure $\mathrm{DM}$ imparts a time delay $\Delta t$ that follows $\Delta t=\mathrm{DM}\ K_\mathrm{DM} (\nu^{-2} - \nu_\mathrm{ref}^{-2})$, where $\nu$ is the observing frequency, and $\nu_\mathrm{ref}$ is a reference frequency. For $\Delta$ in seconds, $\mathrm{DM}$ in pc cm$^{-3}$, and $\nu$ in MHz, the constant is defined as $K_\mathrm{DM}=(2.41\times10^{-4})^{-1}$ MHz$^2$ s cm$^3$ pc$^{-1}$. See Kulkarni 2020 (https://arxiv.org/abs/2007.02886) for a useful derivation.
* Useful `python` functions to use for this challenge are `scipy.stats.sigmaclip`, `np.fft.rfft`, `np.fft.rfftfreq`.
