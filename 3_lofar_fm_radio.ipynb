# Listening to FM radio with LOFAR

While the LOFAR LBA and HBA bands avoid the FM radio stations that transmit between 88 and 108MHz, the signals are still detectable.
![image](img/lofar_spec.png)
In this challenge you are tasked to identify FM signal in LOFAR observations of a single station and demodulate the complex voltages to an audio signal that you can listen to. FM stands for Frequency Modulation and is a scheme to modulate an input signal into variations in transmission frequency, see https://en.wikipedia.org/wiki/Frequency_modulation.

You'll learn to mix, filter resample and visualize LOFAR voltages and convert them to audio signals with standard python libraries (`numpy`, `scipy`, `matplotlib` etc.).

### Hints and tips

* The `scipy.signal` library has several functions that will be useful, e.g. `resample_poly`, `sosfilt`, `butter`.
* The `scipy.io.wavfile` module has functions for writing audio files in the `wav` format.
* Audio files are usually written at 48000 kHz sample rates.
* FM demodulation can be done via different methods. The simplest used here is called *Quadrature demodulation*, which uses the phase difference between adjacent samples to represent the frequency. Use the demodulated signal $x_i=\arg{(e_i e_{i+1}^*)}$, where $e$ is a complex voltage and the index $i$ denotes sample index. $\arg$ denotes the phase of the complex number and $e^*$ indicates complex conjugation. See https://wiki.gnuradio.org/index.php/Quadrature_Demod.
