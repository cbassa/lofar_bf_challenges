# Measuring and correcting for Faraday rotation in pulsar single pulses

Linearly polarized signals undergo Faraday rotation as they travel through the ionized interstellar medium.

In this challenge you are tasked with determining and correcting for the rotation measure of single pulse from a bright radio pulsar using beamformed complex voltage data from the LOFAR Core stations (Observation L2079334).

You'll learn how to channelize voltage data, construct the Stokes parameters, and determine the rotation measure (RM) from the rotation of Stokes Q and U with frequency, and correct the data for it.

### Hints and tips

* The Stokes parameters describe the polarization of a signal, where Stokes I gives the total intensity, Stokes Q and U describe linear polarization, and Stokes V describes circular polarization. For voltages detected with linearly polarized antennas such as those in LOFAR stations, the Stokes parameters are defined, via the IAU/IEEE convention as (see https://ui.adsabs.harvard.edu/abs/2010PASA...27..104V/abstract for more information):
    * $I=<|e_\mathrm{x}|^2+|e_\mathrm{y}|^2>$
    * $Q=<|e_\mathrm{x}|^2-|e_\mathrm{y}|^2>$
    * $U=<2\mathrm{Re}[e_\mathrm{x}e_\mathrm{y}^*]>$
    * $V=<2\mathrm{Im}[e_\mathrm{x}e_\mathrm{y}^*]>$.
      
    Here $e_\mathrm{x}$ and $e_\mathrm{y}$ are the complex voltages of linear polarizations X and Y, and $\mathrm{Re}$ and $\mathrm{Im}$ denote the real and imaginary components of a complex number. Furthermore $|x|$ denotes the absolute value of $x$ and $<x>$ denotes averaging of $x$, and $e^*$ indicates complex conjugation.
* Faraday rotation imparts a position angle $\chi$ in radians following $\chi=\mathrm{RM}\lambda^2$ for a rotation measure $\mathrm{RM}$ in rad m$^{-2}$ and wavelength $\lambda$ in m.
* Furthermore, the position angle $\chi$ is defined as $\tan{2\chi}=\frac{U}{Q}$.
* To speed up calculations, only select the time range covering one bright single pulse.
* Where possible, use data lengths of $2^n$ to allow for easy averaging and reshaping of arrays in `numpy`.
